# Necessary Libraries

```python
import numpy as np
from sympy import Matrix
import matplotlib.pyplot as plt
import seaborn as sns
```

# 1. Create my Personal Matrix


```python
# My id:  0432320005101116
d1 = 6
d2 = 1
A = np.array([
    [d1+2, d2+1], [2*d1, d2+2]
])

print("The Matrix is:\n", A);
```

# 2. Analyze Matrix A


```python
# 1. Size of a matrix
size_A = A.shape  # Returns dimensions (rows, columns)
print("Size of the matrix A:", size_A)

# 2. Determinant of a matrix
det_A = np.linalg.det(A)  # Returns the determinant of the matrix
print("Determinant of the matrix A:", det_A)

# 3. Rank of a matrix
rank_A = np.linalg.matrix_rank(A)  # Returns the rank of the matrix
print("Rank of the matrix A:", rank_A)

# 4. Eigenvalues
eigen_values = np.linalg.eig(A).eigenvalues
print("Eigenvalues of the matrix A:", eigen_values)

# 5. Inverse (only if determinant ≠ 0)
if det_A != 0:
  print("Inverse matrix: \n" , np.linalg.inv(A))
else:
  print("Matrix is singular, cannot compute inverse")

```

# 3. Value Change Experiment

```python
# Matrix B
B = A.copy()

# Add +1 to the element at row 1, column 2
B[0,1] = B[0,1] + 1;
print ("The matrix B is:\n", B)
```

# 4. Re-analyze Matrix B


```python
# 1. Size of a matrix
size_B = B.shape  # Returns dimensions (rows, columns)
print("Size of the matrix B:", size_B)

# 2. Determinant of a matrix
det_B = np.linalg.det(B)  # Returns the determinant of the matrix
print("Determinant of the matrix B:", det_B)

# 3. Rank of a matrix
rank_B = np.linalg.matrix_rank(B)  # Returns the rank of the matrix
print("Rank of the matrix B:", rank_B)

# 4. Eigenvalues
eigen_values = np.linalg.eig(B).eigenvalues
print("Eigenvalues of the matrix B:", eigen_values)

# 5. Inverse (only if determinant ≠ 0)
if det_B != 0:
  print("Inverse matrix: \n" , np.linalg.inv(B))
else:
  print("Matrix is singular, cannot compute inverse")

```

# 5. Comparison & Explanation

# 1. How did the determinant change and why?
It went from 0 to 3. The reason A had a 0 determinant in the first place is that row 2 was just 1.5 times row 1, so the rows weren't really independent. Once I bumped a11 up from 8 to 9, that exact relationship broke, and the determinant moved away from zero.

# 2. Did the rank change? Explain.
Yes, rank went from 1 to 2 (full rank for a 2×2 matrix). Before the change, both rows of A pointed in basically the same direction (row2 = 1.5 × row1), so there was really only one independent row. After changing a11, the rows are no longer proportional, so both count as independent now.

# 3. How did the eigenvalues respond to the value change?
A's eigenvalues were 0 and 11. B's came out to about 0.255 and 11.745. That 0 eigenvalue is basically what made A singular in the first place, so once B stopped being singular, that eigenvalue moved off zero. The bigger eigenvalue barely changed — just a small bump up.

# 4. Is B easier or harder to invert than A? Why?
B is definitely easier — A can't be inverted at all since its determinant is 0, while B has a determinant of 3, so its inverse exists and I could actually compute it. That said, B isn't perfectly well-conditioned either — its smaller eigenvalue (~0.255) is a lot smaller than the other one (~11.745), so it's still somewhat close to singular, just not singular.
